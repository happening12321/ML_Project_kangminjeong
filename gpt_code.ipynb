```python
# ============================================================
# SNP 유전자형 기반 소 개체 분류 및 유사도 분석
# ============================================================

# ============================================================
# 0. 라이브러리
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from IPython.display import display

from scipy.stats import chi2_contingency
from scipy.cluster.hierarchy import linkage, dendrogram

from sklearn.model_selection import (
    train_test_split,
    StratifiedKFold,
    cross_validate
)

from sklearn.preprocessing import OneHotEncoder, LabelEncoder

from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier

from sklearn.metrics import (
    accuracy_score,
    f1_score,
    precision_score,
    recall_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    pairwise_distances
)

from sklearn.decomposition import PCA

from xgboost import XGBClassifier

import time
import warnings

warnings.filterwarnings("ignore")


# ============================================================
# 1. 데이터 불러오기
# ============================================================

df = pd.read_csv("train.csv")
snp_info = pd.read_csv("snp_info.csv")

print("=" * 60)
print("1. 데이터 기본 정보")
print("=" * 60)

print("train 데이터 크기:", df.shape)
print("snp_info 데이터 크기:", snp_info.shape)

print("\n[train 데이터 정보]")
df.info()

print("\n[train 데이터 상위 5개]")
display(df.head())

print("\n[SNP 정보 상위 5개]")
display(snp_info.head())


# ============================================================
# 2. SNP 컬럼 정의
# ============================================================

snp_cols = [f"SNP_{i:02d}" for i in range(1, 16)]

print("\n분석에 사용할 SNP:")
print(snp_cols)


# ============================================================
# 3. 데이터 품질 확인
# ============================================================

print("\n" + "=" * 60)
print("2. 데이터 품질 확인")
print("=" * 60)

# 결측치
missing = df.isnull().sum().to_frame("결측치")
print("\n[전체 결측치]")
print(missing)

# 중복
print("\n전체 중복 행:", df.duplicated().sum())
print("중복 ID:", df["id"].duplicated().sum())

# SNP 결측치
print("\n[SNP 결측치]")
display(missing.loc[snp_cols])


# ============================================================
# 4. Class 분포
# ============================================================

print("\n" + "=" * 60)
print("3. Class 분포")
print("=" * 60)

class_counts = (df["class"].value_counts().sort_index())

class_ratio = (df["class"].value_counts(normalize=True).sort_index().mul(100).round(2))

class_summary = pd.DataFrame({"개체 수": class_counts,"비율(%)": class_ratio})

display(class_summary)

plt.figure(figsize=(6, 4))

class_counts.plot(kind="bar")

plt.title("Class Distribution")
plt.xlabel("Class")
plt.ylabel("Number of Individuals")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# 5. SNP별 유전자형 분포
# ============================================================

print("\n" + "=" * 60)
print("4. SNP별 유전자형 분포")
print("=" * 60)

genotype_summary = pd.DataFrame({
    "SNP": snp_cols,
    "고유 유전자형 수": [
        df[col].nunique()
        for col in snp_cols
    ]
})

display(genotype_summary)

for col in snp_cols:

    print(f"\n[{col}]")
    display(
        df[col]
        .value_counts()
        .to_frame("개체 수")
    )


# ============================================================
# 6. SNP × Class 분포
# ============================================================

print("\n" + "=" * 60)
print("5. SNP × Class 유전자형 분포")
print("=" * 60)

for col in snp_cols:

    table = pd.crosstab(
        df[col],
        df["class"],
        normalize="columns"
    ) * 100

    print(f"\n===== {col} =====")
    display(table.round(2))


# ============================================================
# 7. SNP × Class 시각화
# ============================================================

print("\n" + "=" * 60)
print("6. SNP × Class 시각화")
print("=" * 60)

for col in snp_cols:

    table = pd.crosstab(
        df[col],
        df["class"],
        normalize="columns"
    ) * 100

    ax = table.plot(
        kind="bar",
        figsize=(7, 4)
    )

    ax.set_title(
        f"{col} Genotype Distribution by Class"
    )

    ax.set_xlabel("Genotype")
    ax.set_ylabel("Percentage (%)")
    ax.legend(title="Class")

    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()


# ============================================================
# 8. 카이제곱 검정
# ============================================================

print("\n" + "=" * 60)
print("7. SNP와 Class의 연관성 분석")
print("=" * 60)

chi_results = []

for col in snp_cols:

    table = pd.crosstab(
        df[col],
        df["class"]
    )

    chi2, p_value, dof, expected = chi2_contingency(table)

    chi_results.append({
        "SNP": col,
        "chi2": chi2,
        "p_value": p_value,
        "유의여부": (
            "유의"
            if p_value < 0.05
            else "비유의"
        )
    })

chi_df = pd.DataFrame(chi_results)

chi_df = chi_df.sort_values(
    "p_value"
)

print("\n[SNP × 카이제곱 결과]")
display(
    chi_df.round(5)
)


# ============================================================
# 9. SNP 정보와 카이제곱 결과 결합
# ============================================================

chi_info_df = chi_df.merge(
    snp_info,
    left_on="SNP",
    right_on="SNP_id",
    how="left"
)

print("\n[SNP 정보 + 카이제곱 결과]")
display(chi_info_df.round(5))


# ============================================================
# 10. 분석용 X, y 구성
# ============================================================

print("\n" + "=" * 60)
print("8. 머신러닝 데이터 구성")
print("=" * 60)

X = df[snp_cols].copy()
y = df["class"].copy()

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nX 상위 5개")
display(X.head())

print("\ny 분포")
display(y.value_counts())


# ============================================================
# 11. Train / Test Split
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

print("\n" + "=" * 60)
print("9. Train / Test Split")
print("=" * 60)

print("Train:", X_train.shape)
print("Test :", X_test.shape)

print("\n[Train class 비율]")
print(
    y_train
    .value_counts(normalize=True)
    .sort_index()
    .round(3)
)

print("\n[Test class 비율]")
print(
    y_test
    .value_counts(normalize=True)
    .sort_index()
    .round(3)
)


# ============================================================
# 12. One-Hot Encoding
# ============================================================

print("\n" + "=" * 60)
print("10. One-Hot Encoding")
print("=" * 60)

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

X_train_encoded = encoder.fit_transform(
    X_train
)

X_test_encoded = encoder.transform(
    X_test
)

print("Encoding 전:", X_train.shape)
print("Encoding 후:", X_train_encoded.shape)

feature_names = encoder.get_feature_names_out(
    snp_cols
)

print("\nEncoding된 Feature 수:")
print(len(feature_names))


# ============================================================
# 13. Label Encoding
# ============================================================

label_encoder = LabelEncoder()

y_train_encoded = label_encoder.fit_transform(
    y_train
)

y_test_encoded = label_encoder.transform(
    y_test
)

print("\nClass Encoding:")
for cls, num in zip(
    label_encoder.classes_,
    range(len(label_encoder.classes_))
):
    print(f"{cls} → {num}")


# ============================================================
# 14. Random Forest
# ============================================================

print("\n" + "=" * 60)
print("11. Random Forest")
print("=" * 60)

rf = RandomForestClassifier(
    n_estimators=300,
    max_depth=None,
    min_samples_split=2,
    random_state=42
)

start = time.time()

rf.fit(
    X_train_encoded,
    y_train
)

rf_time = time.time() - start

rf_pred = rf.predict(
    X_test_encoded
)

rf_accuracy = accuracy_score(
    y_test,
    rf_pred
)

rf_f1 = f1_score(
    y_test,
    rf_pred,
    average="macro"
)

rf_precision = precision_score(
    y_test,
    rf_pred,
    average="macro"
)

rf_recall = recall_score(
    y_test,
    rf_pred,
    average="macro"
)

print("Accuracy :", round(rf_accuracy, 4))
print("Macro F1 :", round(rf_f1, 4))
print("Precision:", round(rf_precision, 4))
print("Recall   :", round(rf_recall, 4))
print("학습 시간:", round(rf_time, 4), "초")

print("\n[Classification Report]")
print(
    classification_report(
        y_test,
        rf_pred
    )
)


# ============================================================
# 15. Random Forest Confusion Matrix
# ============================================================

cm_rf = confusion_matrix(
    y_test,
    rf_pred,
    labels=["A", "B", "C"]
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_rf,
    display_labels=["A", "B", "C"]
)

disp.plot()

plt.title("Random Forest Confusion Matrix")
plt.tight_layout()
plt.show()


# ============================================================
# 16. Random Forest Feature Importance
# ============================================================

print("\n" + "=" * 60)
print("12. Random Forest Feature Importance")
print("=" * 60)

rf_feature_importance = pd.DataFrame({
    "feature": feature_names,
    "importance": rf.feature_importances_
})

rf_feature_importance = (
    rf_feature_importance
    .sort_values(
        "importance",
        ascending=False
    )
)

print("\n[상위 20개 유전자형 Feature]")
display(
    rf_feature_importance.head(20)
)


# ============================================================
# 17. SNP 단위 중요도 계산
# ============================================================

snp_importance = []

for snp in snp_cols:

    mask = np.array([
        name.startswith(f"{snp}_")
        for name in feature_names
    ])

    importance_sum = (
        rf.feature_importances_[mask].sum()
    )

    snp_importance.append({
        "SNP": snp,
        "importance": importance_sum
    })

snp_importance_df = pd.DataFrame(
    snp_importance
)

snp_importance_df = (
    snp_importance_df
    .sort_values(
        "importance",
        ascending=False
    )
)

snp_importance_df = snp_importance_df.merge(
    snp_info,
    left_on="SNP",
    right_on="SNP_id",
    how="left"
)

print("\n[SNP별 Feature Importance]")
display(
    snp_importance_df.round(5)
)


# ============================================================
# 18. SNP Importance 시각화
# ============================================================

plt.figure(figsize=(9, 6))

plot_df = (
    snp_importance_df
    .sort_values("importance")
)

plt.barh(
    plot_df["SNP"],
    plot_df["importance"]
)

plt.xlabel("Importance")
plt.ylabel("SNP")
plt.title("Random Forest SNP Feature Importance")

plt.tight_layout()
plt.show()


# ============================================================
# 19. XGBoost
# ============================================================

print("\n" + "=" * 60)
print("13. XGBoost")
print("=" * 60)

xgb = XGBClassifier(
    n_estimators=300,
    max_depth=3,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="multi:softprob",
    num_class=3,
    eval_metric="mlogloss",
    random_state=42
)

start = time.time()

xgb.fit(
    X_train_encoded,
    y_train_encoded
)

xgb_time = time.time() - start

xgb_pred = xgb.predict(
    X_test_encoded
)

xgb_pred_class = (
    label_encoder
    .inverse_transform(xgb_pred)
)

xgb_accuracy = accuracy_score(
    y_test,
    xgb_pred_class
)

xgb_f1 = f1_score(
    y_test,
    xgb_pred_class,
    average="macro"
)

xgb_precision = precision_score(
    y_test,
    xgb_pred_class,
    average="macro"
)

xgb_recall = recall_score(
    y_test,
    xgb_pred_class,
    average="macro"
)

print("Accuracy :", round(xgb_accuracy, 4))
print("Macro F1 :", round(xgb_f1, 4))
print("Precision:", round(xgb_precision, 4))
print("Recall   :", round(xgb_recall, 4))
print("학습 시간:", round(xgb_time, 4), "초")

print("\n[Classification Report]")
print(
    classification_report(
        y_test,
        xgb_pred_class
    )
)


# ============================================================
# 20. XGBoost Confusion Matrix
# ============================================================

cm_xgb = confusion_matrix(
    y_test,
    xgb_pred_class,
    labels=["A", "B", "C"]
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_xgb,
    display_labels=["A", "B", "C"]
)

disp.plot()

plt.title("XGBoost Confusion Matrix")
plt.tight_layout()
plt.show()


# ============================================================
# 21. XGBoost Feature Importance
# ============================================================

xgb_importance = pd.DataFrame({
    "feature": feature_names,
    "importance": xgb.feature_importances_
})

xgb_importance = (
    xgb_importance
    .sort_values(
        "importance",
        ascending=False
    )
)

print("\n[XGBoost 상위 20개 Feature]")
display(
    xgb_importance.head(20)
)


# ============================================================
# 22. XGBoost SNP 단위 중요도
# ============================================================

xgb_snp_importance = []

for snp in snp_cols:

    mask = np.array([
        name.startswith(f"{snp}_")
        for name in feature_names
    ])

    importance_sum = (
        xgb.feature_importances_[mask].sum()
    )

    xgb_snp_importance.append({
        "SNP": snp,
        "importance": importance_sum
    })

xgb_snp_importance_df = pd.DataFrame(
    xgb_snp_importance
)

xgb_snp_importance_df = (
    xgb_snp_importance_df
    .sort_values(
        "importance",
        ascending=False
    )
)

xgb_snp_importance_df = (
    xgb_snp_importance_df
    .merge(
        snp_info,
        left_on="SNP",
        right_on="SNP_id",
        how="left"
    )
)

print("\n[XGBoost SNP별 Feature Importance]")
display(
    xgb_snp_importance_df.round(5)
)


# ============================================================
# 23. KNN K값 비교
# ============================================================

print("\n" + "=" * 60)
print("14. KNN K값 비교")
print("=" * 60)

k_results = []

for k in range(1, 21):

    knn_temp = KNeighborsClassifier(
        n_neighbors=k,
        metric="hamming",
        weights="distance"
    )

    knn_temp.fit(
        X_train_encoded,
        y_train
    )

    pred = knn_temp.predict(
        X_test_encoded
    )

    acc = accuracy_score(
        y_test,
        pred
    )

    f1 = f1_score(
        y_test,
        pred,
        average="macro"
    )

    k_results.append({
        "K": k,
        "Accuracy": acc,
        "Macro F1": f1
    })

k_df = pd.DataFrame(k_results)

display(
    k_df.round(4)
)


# ============================================================
# 24. KNN 성능 그래프
# ============================================================

plt.figure(figsize=(8, 5))

plt.plot(
    k_df["K"],
    k_df["Accuracy"],
    marker="o",
    label="Accuracy"
)

plt.plot(
    k_df["K"],
    k_df["Macro F1"],
    marker="o",
    label="Macro F1"
)

plt.xlabel("K")
plt.ylabel("Score")
plt.title("KNN Performance by K")

plt.legend()
plt.grid()

plt.tight_layout()
plt.show()


# ============================================================
# 25. 최적 K 선택
# ============================================================

best_k = int(
    k_df.loc[
        k_df["Macro F1"].idxmax(),
        "K"
    ]
)

print("Macro F1 기준 최적 K:", best_k)


# ============================================================
# 26. 최적 KNN
# ============================================================

print("\n" + "=" * 60)
print("15. 최적 KNN")
print("=" * 60)

knn = KNeighborsClassifier(
    n_neighbors=best_k,
    metric="hamming",
    weights="distance"
)

start = time.time()

knn.fit(
    X_train_encoded,
    y_train
)

knn_time = time.time() - start

knn_pred = knn.predict(
    X_test_encoded
)

knn_accuracy = accuracy_score(
    y_test,
    knn_pred
)

knn_f1 = f1_score(
    y_test,
    knn_pred,
    average="macro"
)

knn_precision = precision_score(
    y_test,
    knn_pred,
    average="macro"
)

knn_recall = recall_score(
    y_test,
    knn_pred,
    average="macro"
)

print("최적 K   :", best_k)
print("Accuracy :", round(knn_accuracy, 4))
print("Macro F1 :", round(knn_f1, 4))
print("Precision:", round(knn_precision, 4))
print("Recall   :", round(knn_recall, 4))
print("학습 시간:", round(knn_time, 4), "초")

print("\n[Classification Report]")
print(
    classification_report(
        y_test,
        knn_pred
    )
)


# ============================================================
# 27. KNN Confusion Matrix
# ============================================================

cm_knn = confusion_matrix(
    y_test,
    knn_pred,
    labels=["A", "B", "C"]
)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm_knn,
    display_labels=["A", "B", "C"]
)

disp.plot()

plt.title("KNN Confusion Matrix")
plt.tight_layout()
plt.show()


# ============================================================
# 28. Cross Validation
# ============================================================

print("\n" + "=" * 60)
print("16. Stratified 5-Fold Cross Validation")
print("=" * 60)

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


# ------------------------------------------------------------
# Random Forest CV
# ------------------------------------------------------------

rf_cv = cross_validate(
    rf,
    X_train_encoded,
    y_train,
    cv=skf,
    scoring=[
        "accuracy",
        "f1_macro"
    ],
    return_train_score=False
)

rf_cv_accuracy = (
    rf_cv["test_accuracy"].mean()
)

rf_cv_f1 = (
    rf_cv["test_f1_macro"].mean()
)


# ------------------------------------------------------------
# XGBoost CV
# ------------------------------------------------------------

xgb_cv = cross_validate(
    xgb,
    X_train_encoded,
    y_train_encoded,
    cv=skf,
    scoring=[
        "accuracy",
        "f1_macro"
    ],
    return_train_score=False
)

xgb_cv_accuracy = (
    xgb_cv["test_accuracy"].mean()
)

xgb_cv_f1 = (
    xgb_cv["test_f1_macro"].mean()
)


# ------------------------------------------------------------
# KNN CV
# ------------------------------------------------------------

knn_cv = cross_validate(
    knn,
    X_train_encoded,
    y_train,
    cv=skf,
    scoring=[
        "accuracy",
        "f1_macro"
    ],
    return_train_score=False
)

knn_cv_accuracy = (
    knn_cv["test_accuracy"].mean()
)

knn_cv_f1 = (
    knn_cv["test_f1_macro"].mean()
)


# ============================================================
# 29. CV 결과
# ============================================================

cv_results = pd.DataFrame({
    "모델": [
        "Random Forest",
        "XGBoost",
        "KNN"
    ],
    "CV Accuracy": [
        rf_cv_accuracy,
        xgb_cv_accuracy,
        knn_cv_accuracy
    ],
    "CV Macro F1": [
        rf_cv_f1,
        xgb_cv_f1,
        knn_cv_f1
    ]
})

display(
    cv_results.round(4)
)


# ============================================================
# 30. 최종 모델 성능 비교
# ============================================================

print("\n" + "=" * 60)
print("17. 최종 모델 성능 비교")
print("=" * 60)

final_results = pd.DataFrame({
    "모델": [
        "Random Forest",
        "XGBoost",
        "KNN"
    ],

    "CV Accuracy": [
        rf_cv_accuracy,
        xgb_cv_accuracy,
        knn_cv_accuracy
    ],

    "CV Macro F1": [
        rf_cv_f1,
        xgb_cv_f1,
        knn_cv_f1
    ],

    "Test Accuracy": [
        rf_accuracy,
        xgb_accuracy,
        knn_accuracy
    ],

    "Test Macro F1": [
        rf_f1,
        xgb_f1,
        knn_f1
    ],

    "Precision": [
        rf_precision,
        xgb_precision,
        knn_precision
    ],

    "Recall": [
        rf_recall,
        xgb_recall,
        knn_recall
    ],

    "학습시간(초)": [
        rf_time,
        xgb_time,
        knn_time
    ]
})

display(
    final_results.round(4)
)


# ============================================================
# 31. Macro F1 기준 최종 모델 확인
# ============================================================

best_model_name = (
    final_results
    .loc[
        final_results["CV Macro F1"].idxmax(),
        "모델"
    ]
)

print(
    "CV Macro F1 기준 최고 성능 모델:",
    best_model_name
)


# ============================================================
# 32. PCA
# ============================================================

print("\n" + "=" * 60)
print("18. PCA 분석")
print("=" * 60)

# 전체 데이터 Encoding
X_all_encoded = encoder.fit_transform(
    df[snp_cols]
)

pca = PCA(
    n_components=2,
    random_state=42
)

X_pca = pca.fit_transform(
    X_all_encoded
)

pca_df = pd.DataFrame({
    "PC1": X_pca[:, 0],
    "PC2": X_pca[:, 1],
    "class": df["class"],
    "id": df["id"]
})

pc1_ratio = (
    pca.explained_variance_ratio_[0] * 100
)

pc2_ratio = (
    pca.explained_variance_ratio_[1] * 100
)

total_ratio = (
    pca.explained_variance_ratio_.sum() * 100
)

print(
    "PC1 설명분산:",
    round(pc1_ratio, 2),
    "%"
)

print(
    "PC2 설명분산:",
    round(pc2_ratio, 2),
    "%"
)

print(
    "PC1 + PC2 설명분산:",
    round(total_ratio, 2),
    "%"
)


# ============================================================
# 33. PCA 시각화
# ============================================================

plt.figure(figsize=(8, 6))

for cls in ["A", "B", "C"]:

    subset = pca_df[
        pca_df["class"] == cls
    ]

    plt.scatter(
        subset["PC1"],
        subset["PC2"],
        label=cls
    )

plt.xlabel(
    f"PC1 ({pc1_ratio:.2f}%)"
)

plt.ylabel(
    f"PC2 ({pc2_ratio:.2f}%)"
)

plt.title("PCA of SNP Genotypes")

plt.legend()
plt.grid()

plt.tight_layout()
plt.show()


# ============================================================
# 34. 개체 간 Hamming Distance
# ============================================================

print("\n" + "=" * 60)
print("19. 개체 간 유전자형 거리")
print("=" * 60)

distance_matrix = pairwise_distances(
    X_all_encoded,
    metric="hamming"
)

distance_df = pd.DataFrame(
    distance_matrix,
    index=df["id"],
    columns=df["id"]
)

print("Distance Matrix 크기:")
print(distance_df.shape)

display(
    distance_df.iloc[:5, :5]
)


# ============================================================
# 35. 거리 Heatmap
# ============================================================

plt.figure(figsize=(10, 8))

plt.imshow(
    distance_matrix,
    aspect="auto"
)

plt.colorbar(
    label="Hamming Distance"
)

plt.title(
    "Individual Genotype Distance"
)

plt.xlabel("Individuals")
plt.ylabel("Individuals")

plt.tight_layout()
plt.show()


# ============================================================
# 36. 특정 개체 유사 개체 탐색 함수
# ============================================================

def find_similar_individuals(
    target_id,
    distance_df,
    df,
    n=5
):

    distances = (
        distance_df
        .loc[target_id]
        .drop(target_id)
    )

    result = pd.DataFrame({
        "id": distances.index,
        "distance": distances.values
    })

    result = (
        result
        .sort_values("distance")
        .head(n)
    )

    result = result.merge(
        df[["id", "class"]],
        on="id",
        how="left"
    )

    result["similarity(%)"] = (
        (1 - result["distance"]) * 100
    ).round(2)

    return result


# ============================================================
# 37. TRAIN_001 유사 개체 탐색
# ============================================================

print("\n" + "=" * 60)
print("20. TRAIN_001 유사 개체 탐색")
print("=" * 60)

target_id = "TRAIN_001"

target_class = df.loc[
    df["id"] == target_id,
    "class"
].iloc[0]

print("대상 개체:", target_id)
print("실제 class:", target_class)

similar_train001 = find_similar_individuals(
    target_id,
    distance_df,
    df,
    n=5
)

display(similar_train001)


# ============================================================
# 38. Top-N 동일 Class 비율
# ============================================================

print("\n" + "=" * 60)
print("21. Top-N 유사 개체의 동일 Class 비율")
print("=" * 60)

topn_results = []

df_id_indexed = df.set_index("id")

for target_id in df["id"]:

    target_class = (
        df_id_indexed
        .loc[target_id, "class"]
    )

    distances = (
        distance_df
        .loc[target_id]
        .drop(target_id)
        .sort_values()
    )

    for n in [1, 3, 5]:

        top_ids = distances.index[:n]

        top_classes = (
            df_id_indexed
            .loc[top_ids, "class"]
        )

        same_class_ratio = (
            top_classes == target_class
        ).mean()

        topn_results.append({
            "target_id": target_id,
            "target_class": target_class,
            "N": n,
            "same_class_ratio": same_class_ratio
        })


topn_df = pd.DataFrame(
    topn_results
)

display(
    topn_df.head(10)
)


# ============================================================
# 39. Top-N 평균 동일 Class 비율
# ============================================================

topn_summary = (
    topn_df
    .groupby("N")["same_class_ratio"]
    .mean()
    .mul(100)
    .round(2)
    .reset_index()
)

topn_summary.columns = [
    "Top-N",
    "동일 class 비율(%)"
]

print("\n[Top-N 평균 동일 class 비율]")
display(topn_summary)


# ============================================================
# 40. Top-N 동일 Class 비율 시각화
# ============================================================

plt.figure(figsize=(7, 4))

plt.bar(
    topn_summary["Top-N"].astype(str),
    topn_summary["동일 class 비율(%)"]
)

plt.xlabel("Top-N")
plt.ylabel("Same Class Ratio (%)")
plt.title(
    "Same Class Ratio among Similar Individuals"
)

plt.tight_layout()
plt.show()


# ============================================================
# 41. Hierarchical Clustering
# ============================================================

print("\n" + "=" * 60)
print("22. Hierarchical Clustering")
print("=" * 60)

Z = linkage(
    X_all_encoded,
    method="average",
    metric="hamming"
)

plt.figure(figsize=(14, 6))

dendrogram(
    Z,
    labels=df["id"].values,
    leaf_rotation=90,
    leaf_font_size=7
)

plt.title(
    "Hierarchical Clustering of Individuals"
)

plt.xlabel("Individual ID")
plt.ylabel("Hamming Distance")

plt.tight_layout()
plt.show()


# ============================================================
# 42. 최종 결과 요약
# ============================================================

print("\n" + "=" * 60)
print("23. 프로젝트 최종 결과 요약")
print("=" * 60)

print("\n[데이터]")
print("개체 수:", len(df))
print("SNP 수:", len(snp_cols))
print("Class:", sorted(df["class"].unique()))

print("\n[Class 분포]")
display(class_summary)

print("\n[모델 성능]")
display(final_results.round(4))

print("\n[최고 성능 모델]")
print(best_model_name)

print("\n[Top-N 유사 개체 동일 Class 비율]")
display(topn_summary)

print("\n[PCA]")
print(
    f"PC1 설명분산: {pc1_ratio:.2f}%"
)
print(
    f"PC2 설명분산: {pc2_ratio:.2f}%"
)
print(
    f"PC1+PC2 설명분산: {total_ratio:.2f}%"
)

print("\n[Random Forest 중요 SNP Top 5]")
display(
    snp_importance_df[
        [
            "SNP",
            "name",
            "chrom",
            "cm",
            "pos",
            "importance"
        ]
    ].head(5).round(5)
)

print("\n분석 완료!")
```

```python
# ============================================================
# SNP 유전자형 기반 소 개체 Class 분류 및 유사성 분석
# ============================================================

# ============================================================
# 0. 라이브러리
# ============================================================

import os
import time
import warnings

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split, StratifiedKFold, cross_validate
from sklearn.preprocessing import OneHotEncoder, LabelEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline

from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.decomposition import PCA

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay
)

from xgboost import XGBClassifier

warnings.filterwarnings("ignore")

# 그래프 한글 설정
plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False

RANDOM_STATE = 42


# ============================================================
# 1. 데이터 불러오기
# ============================================================

df = pd.read_csv("train.csv")
snp_info = pd.read_csv("snp_info.csv")

print("=" * 70)
print("1. 데이터 기본 정보")
print("=" * 70)

print(f"전체 개체 수: {len(df):,}")
print(f"전체 컬럼 수: {df.shape[1]:,}")

print("\n[train.csv 컬럼]")
print(df.columns.tolist())

print("\n[snp_info.csv]")
display(snp_info)


# ============================================================
# 2. 기본 데이터 확인
# ============================================================

print("\n" + "=" * 70)
print("2. 데이터 구조 확인")
print("=" * 70)

display(df.head())

print("\n[데이터 타입]")
print(df.dtypes)

print("\n[결측치]")
missing_df = pd.DataFrame({
    "결측치": df.isnull().sum(),
    "결측률(%)": df.isnull().mean() * 100
})

display(missing_df)

print("\n[중복 행]")
print(df.duplicated().sum())


# ============================================================
# 3. SNP 컬럼 지정
# ============================================================

snp_cols = [f"SNP_{i:02d}" for i in range(1, 16)]

print("\nSNP 컬럼:")
print(snp_cols)

# 존재 여부 확인
missing_snp_cols = [col for col in snp_cols if col not in df.columns]

if missing_snp_cols:
    raise ValueError(f"다음 SNP 컬럼이 없습니다: {missing_snp_cols}")


# ============================================================
# 4. SNP 유전자형 정제
# ============================================================

print("\n" + "=" * 70)
print("3. SNP 유전자형 정제")
print("=" * 70)

# 예:
# "G G" → "GG"
# "A G" → "AG"
# "G A" → "GA"

for col in snp_cols:
    df[col] = (
        df[col]
        .astype(str)
        .str.replace(" ", "", regex=False)
        .str.upper()
    )

print("SNP 공백 제거 및 대문자 변환 완료")


# ============================================================
# 5. SNP 유전자형 종류 확인
# ============================================================

print("\n" + "=" * 70)
print("4. SNP별 유전자형 종류")
print("=" * 70)

for col in snp_cols:
    values = sorted(df[col].dropna().unique())

    print(f"{col}: {values}")


# ============================================================
# 6. Class 분포
# ============================================================

print("\n" + "=" * 70)
print("5. Class 분포")
print("=" * 70)

class_counts = df["class"].value_counts().sort_index()
class_ratio = df["class"].value_counts(normalize=True).sort_index() * 100

class_distribution = pd.DataFrame({
    "개체 수": class_counts,
    "비율(%)": class_ratio.round(2)
})

display(class_distribution)


# Class 분포 시각화
plt.figure(figsize=(7, 5))

class_counts.plot(kind="bar")

plt.title("Class별 개체 분포")
plt.xlabel("Class")
plt.ylabel("개체 수")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()


# ============================================================
# 7. SNP별 전체 유전자형 분포
# ============================================================

print("\n" + "=" * 70)
print("6. SNP별 유전자형 분포")
print("=" * 70)

for col in snp_cols:

    genotype_counts = df[col].value_counts()

    plt.figure(figsize=(7, 4))

    genotype_counts.plot(kind="bar")

    plt.title(f"{col} 유전자형 분포")
    plt.xlabel("유전자형")
    plt.ylabel("개체 수")
    plt.xticks(rotation=0)

    plt.tight_layout()
    plt.show()


# ============================================================
# 8. SNP × Class 비율 분석
# ============================================================

print("\n" + "=" * 70)
print("7. SNP별 Class 유전자형 분포")
print("=" * 70)

# 각 SNP에 대해
# 행 = 유전자형
# 열 = Class
# 값 = 비율

for col in snp_cols:

    table = pd.crosstab(
        df[col],
        df["class"],
        normalize="columns"
    ) * 100

    print(f"\n[{col}]")
    display(table.round(2))


# ============================================================
# 9. SNP × Class 히트맵
# ============================================================

print("\n" + "=" * 70)
print("8. SNP × Class 비율 시각화")
print("=" * 70)

# 각 SNP의 유전자형별 Class 비율을 하나의 표로 만들기

heatmap_data = []

for col in snp_cols:

    crosstab = pd.crosstab(
        df[col],
        df["class"],
        normalize="columns"
    ) * 100

    for genotype in crosstab.index:

        row = {
            "SNP": col,
            "Genotype": genotype
        }

        for cls in sorted(df["class"].unique()):
            row[cls] = crosstab.loc[genotype, cls] if cls in crosstab.columns else 0

        heatmap_data.append(row)

heatmap_df = pd.DataFrame(heatmap_data)

display(heatmap_df.head(20))


# ============================================================
# 10. 머신러닝용 X / y 정의
# ============================================================

print("\n" + "=" * 70)
print("9. 머신러닝 데이터 구성")
print("=" * 70)

X = df[snp_cols].copy()
y = df["class"].copy()

print("X 크기:", X.shape)
print("y 크기:", y.shape)

print("\nX:")
display(X.head())

print("\ny:")
display(y.head())


# ============================================================
# 11. Class Label Encoding
# ============================================================

label_encoder = LabelEncoder()

y_encoded = label_encoder.fit_transform(y)

print("\nClass 인코딩:")
for original, encoded in zip(
    label_encoder.classes_,
    label_encoder.transform(label_encoder.classes_)
):
    print(f"{original} → {encoded}")


# ============================================================
# 12. Train / Test 분할
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y_encoded,
    test_size=0.2,
    stratify=y_encoded,
    random_state=RANDOM_STATE
)

print("\n" + "=" * 70)
print("10. 데이터 분할")
print("=" * 70)

print("Train:", X_train.shape)
print("Test :", X_test.shape)


# ============================================================
# 13. SNP One-Hot Encoding
# ============================================================

encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)

X_train_encoded = encoder.fit_transform(X_train)
X_test_encoded = encoder.transform(X_test)

print("\n인코딩 전:", X_train.shape)
print("인코딩 후:", X_train_encoded.shape)


# ============================================================
# 14. Random Forest
# ============================================================

print("\n" + "=" * 70)
print("11. Random Forest")
print("=" * 70)

rf_model = RandomForestClassifier(
    n_estimators=300,
    random_state=RANDOM_STATE,
    class_weight="balanced",
    n_jobs=-1
)

start_time = time.time()

rf_model.fit(
    X_train_encoded,
    y_train
)

rf_train_time = time.time() - start_time

rf_pred = rf_model.predict(X_test_encoded)

rf_accuracy = accuracy_score(y_test, rf_pred)
rf_precision = precision_score(
    y_test,
    rf_pred,
    average="macro",
    zero_division=0
)
rf_recall = recall_score(
    y_test,
    rf_pred,
    average="macro",
    zero_division=0
)
rf_f1 = f1_score(
    y_test,
    rf_pred,
    average="macro",
    zero_division=0
)

print(f"Accuracy : {rf_accuracy:.4f}")
print(f"Precision: {rf_precision:.4f}")
print(f"Recall   : {rf_recall:.4f}")
print(f"Macro F1 : {rf_f1:.4f}")
print(f"학습 시간: {rf_train_time:.4f}초")


# ============================================================
# 15. XGBoost
# ============================================================

print("\n" + "=" * 70)
print("12. XGBoost")
print("=" * 70)

xgb_model = XGBClassifier(
    n_estimators=300,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="multi:softprob",
    num_class=len(label_encoder.classes_),
    eval_metric="mlogloss",
    random_state=RANDOM_STATE,
    n_jobs=-1
)

start_time = time.time()

xgb_model.fit(
    X_train_encoded,
    y_train
)

xgb_train_time = time.time() - start_time

xgb_pred = xgb_model.predict(X_test_encoded)

xgb_accuracy = accuracy_score(y_test, xgb_pred)

xgb_precision = precision_score(
    y_test,
    xgb_pred,
    average="macro",
    zero_division=0
)

xgb_recall = recall_score(
    y_test,
    xgb_pred,
    average="macro",
    zero_division=0
)

xgb_f1 = f1_score(
    y_test,
    xgb_pred,
    average="macro",
    zero_division=0
)

print(f"Accuracy : {xgb_accuracy:.4f}")
print(f"Precision: {xgb_precision:.4f}")
print(f"Recall   : {xgb_recall:.4f}")
print(f"Macro F1 : {xgb_f1:.4f}")
print(f"학습 시간: {xgb_train_time:.4f}초")


# ============================================================
# 16. KNN
# ============================================================

print("\n" + "=" * 70)
print("13. KNN")
print("=" * 70)

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_encoded)
X_test_scaled = scaler.transform(X_test_encoded)

knn_model = KNeighborsClassifier(
    n_neighbors=5
)

start_time = time.time()

knn_model.fit(
    X_train_scaled,
    y_train
)

knn_train_time = time.time() - start_time

knn_pred = knn_model.predict(X_test_scaled)

knn_accuracy = accuracy_score(y_test, knn_pred)

knn_precision = precision_score(
    y_test,
    knn_pred,
    average="macro",
    zero_division=0
)

knn_recall = recall_score(
    y_test,
    knn_pred,
    average="macro",
    zero_division=0
)

knn_f1 = f1_score(
    y_test,
    knn_pred,
    average="macro",
    zero_division=0
)

print(f"Accuracy : {knn_accuracy:.4f}")
print(f"Precision: {knn_precision:.4f}")
print(f"Recall   : {knn_recall:.4f}")
print(f"Macro F1 : {knn_f1:.4f}")
print(f"학습 시간: {knn_train_time:.4f}초")


# ============================================================
# 17. 모델 성능 비교
# ============================================================

print("\n" + "=" * 70)
print("14. 모델 성능 비교")
print("=" * 70)

results = pd.DataFrame({
    "Model": [
        "Random Forest",
        "XGBoost",
        "KNN"
    ],
    "Accuracy": [
        rf_accuracy,
        xgb_accuracy,
        knn_accuracy
    ],
    "Precision": [
        rf_precision,
        xgb_precision,
        knn_precision
    ],
    "Recall": [
        rf_recall,
        xgb_recall,
        knn_recall
    ],
    "Macro F1": [
        rf_f1,
        xgb_f1,
        knn_f1
    ],
    "Training Time": [
        rf_train_time,
        xgb_train_time,
        knn_train_time
    ]
})

from IPython.display import display, HTML

display(
    HTML(
        results.to_html(
            index=False,
            formatters={
                "Accuracy": lambda x: f"{x:.4f}",
                "Precision": lambda x: f"{x:.4f}",
                "Recall": lambda x: f"{x:.4f}",
                "F1": lambda x: f"{x:.4f}"
            }
        )
    )
)


# ============================================================
# 18. 최종 모델 선정
# ============================================================

best_model_name = results.loc[
    results["Macro F1"].idxmax(),
    "Model"
]

print("\n최종 선택 모델:", best_model_name)


# ============================================================
# 19. Random Forest Classification Report
# ============================================================

print("\n" + "=" * 70)
print("15. Random Forest 상세 평가")
print("=" * 70)

print(
    classification_report(
        y_test,
        rf_pred,
        target_names=label_encoder.classes_,
        digits=4
    )
)


# ============================================================
# 20. Confusion Matrix
# ============================================================

cm = confusion_matrix(y_test, rf_pred)

plt.figure(figsize=(6, 5))

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=label_encoder.classes_
)

disp.plot(
    cmap="Blues",
    values_format="d"
)

plt.title("Random Forest Confusion Matrix")
plt.tight_layout()
plt.show()


# ============================================================
# 21. Random Forest Feature Importance
# ============================================================

print("\n" + "=" * 70)
print("16. SNP 중요도 분석")
print("=" * 70)

feature_names = encoder.get_feature_names_out(snp_cols)

importance_df = pd.DataFrame({
    "Encoded Feature": feature_names,
    "Importance": rf_model.feature_importances_
})

importance_df = importance_df.sort_values(
    "Importance",
    ascending=False
)

display(importance_df.head(20))


# ============================================================
# 22. SNP 단위 중요도 집계
# ============================================================

snp_importance = []

for snp in snp_cols:

    mask = importance_df["Encoded Feature"].str.startswith(
        f"{snp}_"
    )

    total_importance = importance_df.loc[
        mask,
        "Importance"
    ].sum()

    snp_importance.append({
        "SNP": snp,
        "Importance": total_importance
    })

snp_importance_df = pd.DataFrame(
    snp_importance
).sort_values(
    "Importance",
    ascending=False
)

print("\nSNP 단위 중요도:")
display(snp_importance_df)


# ============================================================
# 23. SNP 중요도 그래프
# ============================================================

plt.figure(figsize=(9, 6))

plot_df = snp_importance_df.sort_values(
    "Importance",
    ascending=True
)

plt.barh(
    plot_df["SNP"],
    plot_df["Importance"]
)

plt.title("SNP별 Random Forest 중요도")
plt.xlabel("Importance")
plt.ylabel("SNP")

plt.tight_layout()
plt.show()


# ============================================================
# 24. SNP 정보와 중요도 결합
# ============================================================

snp_result = snp_importance_df.merge(
    snp_info,
    left_on="SNP",
    right_on="SNP_id",
    how="left"
)

snp_result = snp_result[
    [
        "SNP",
        "name",
        "chrom",
        "cm",
        "pos",
        "Importance"
    ]
]

snp_result = snp_result.sort_values(
    "Importance",
    ascending=False
)

print("\n" + "=" * 70)
print("17. SNP 중요도 + 유전체 위치")
print("=" * 70)

display(snp_result)


# ============================================================
# 25. PCA
# ============================================================

print("\n" + "=" * 70)
print("18. PCA 분석")
print("=" * 70)

# 전체 데이터를 PCA에 사용
X_all_encoded = encoder.transform(X)

scaler_pca = StandardScaler()

X_all_scaled = scaler_pca.fit_transform(
    X_all_encoded
)

pca = PCA(
    n_components=2
)

X_pca = pca.fit_transform(
    X_all_scaled
)

pca_df = pd.DataFrame({
    "PC1": X_pca[:, 0],
    "PC2": X_pca[:, 1],
    "Class": df["class"].values
})

print(
    f"PC1 설명 분산: {pca.explained_variance_ratio_[0] * 100:.2f}%"
)

print(
    f"PC2 설명 분산: {pca.explained_variance_ratio_[1] * 100:.2f}%"
)

print(
    f"PC1 + PC2 설명 분산: "
    f"{pca.explained_variance_ratio_.sum() * 100:.2f}%"
)


# ============================================================
# 26. PCA 시각화
# ============================================================

plt.figure(figsize=(8, 6))

for cls in sorted(pca_df["Class"].unique()):

    subset = pca_df[
        pca_df["Class"] == cls
    ]

    plt.scatter(
        subset["PC1"],
        subset["PC2"],
        label=f"Class {cls}",
        alpha=0.7
    )

plt.xlabel(
    f"PC1 ({pca.explained_variance_ratio_[0] * 100:.2f}%)"
)

plt.ylabel(
    f"PC2 ({pca.explained_variance_ratio_[1] * 100:.2f}%)"
)

plt.title("SNP 유전자형 PCA")
plt.legend()

plt.tight_layout()
plt.show()


# ============================================================
# 27. 개체 간 SNP 유사도 분석
# ============================================================

print("\n" + "=" * 70)
print("19. 개체 간 SNP 유사도 분석")
print("=" * 70)

# SNP 유전자형을 단순한 범주형 일치 여부로 비교
#
# 예:
# 개체 A = GG AG AA
# 개체 B = GG AG GA
#
# 3개 중 2개가 동일 → 유사도 66.7%

X_genotype = df[snp_cols].values

similarity_results = []

for i in range(len(df)):

    # 자기 자신 제외
    other_indices = [
        j for j in range(len(df))
        if j != i
    ]

    similarities = []

    for j in other_indices:

        same_count = np.sum(
            X_genotype[i] == X_genotype[j]
        )

        similarity = same_count / len(snp_cols)

        similarities.append(
            (j, similarity)
        )

    # 유사도 높은 순
    similarities = sorted(
        similarities,
        key=lambda x: x[1],
        reverse=True
    )

    similarity_results.append(
        similarities
    )


# ============================================================
# 28. Top 1 / 3 / 5 유사 개체 동일 Class 비율
# ============================================================

top_n_results = []

for n in [1, 3, 5]:

    same_class_count = 0
    total_count = 0

    for i in range(len(df)):

        top_similar = similarity_results[i][:n]

        target_class = df.iloc[i]["class"]

        similar_classes = [
            df.iloc[j]["class"]
            for j, similarity in top_similar
        ]

        same_class_count += sum(
            cls == target_class
            for cls in similar_classes
        )

        total_count += len(similar_classes)

    ratio = (
        same_class_count /
        total_count *
        100
    )

    top_n_results.append({
        "Top N": n,
        "동일 Class 개체 수": same_class_count,
        "전체 비교 개체 수": total_count,
        "동일 Class 비율(%)": ratio
    })

similarity_summary = pd.DataFrame(
    top_n_results
)

print("\n유사 개체의 동일 Class 비율:")
display(
    similarity_summary.round(2)
)


# ============================================================
# 29. 개별 개체 유사도 결과
# ============================================================

def get_similar_individuals(index, top_n=5):

    results = []

    for j, similarity in similarity_results[index][:top_n]:

        results.append({
            "Target ID": df.iloc[index]["id"],
            "Target Class": df.iloc[index]["class"],
            "Similar ID": df.iloc[j]["id"],
            "Similar Class": df.iloc[j]["class"],
            "SNP Similarity (%)": similarity * 100
        })

    return pd.DataFrame(results)


# 예시: 첫 번째 개체
display(
    get_similar_individuals(
        index=0,
        top_n=5
    )
)


# ============================================================
# 30. Class별 평균 유사도
# ============================================================

class_similarity = []

for i in range(len(df)):

    top5 = similarity_results[i][:5]

    mean_similarity = np.mean([
        similarity
        for _, similarity in top5
    ])

    class_similarity.append({
        "ID": df.iloc[i]["id"],
        "Class": df.iloc[i]["class"],
        "Top5 평균 유사도": mean_similarity
    })

class_similarity_df = pd.DataFrame(
    class_similarity
)

class_similarity_summary = (
    class_similarity_df
    .groupby("Class")["Top5 평균 유사도"]
    .mean()
    .reset_index()
)

class_similarity_summary[
    "Top5 평균 유사도(%)"
] = (
    class_similarity_summary[
        "Top5 평균 유사도"
    ] * 100
)

print("\nClass별 Top5 평균 유사도:")
display(
    class_similarity_summary[
        ["Class", "Top5 평균 유사도(%)"]
    ].round(2)
)


# ============================================================
# 31. 최종 결과 요약
# ============================================================

print("\n")
print("=" * 70)
print("최종 프로젝트 결과 요약")
print("=" * 70)

print(f"전체 개체 수: {len(df)}")
print(f"SNP 수: {len(snp_cols)}")
print(f"Class: {list(label_encoder.classes_)}")

print("\n[Class 분포]")
print(class_distribution)

print("\n[모델 성능]")

display(
    HTML(
        results.to_html(
            index=False,
            formatters={
                "Accuracy": lambda x: f"{x:.4f}",
                "Precision": lambda x: f"{x:.4f}",
                "Recall": lambda x: f"{x:.4f}",
                "F1": lambda x: f"{x:.4f}"
            }
        )
    )
)

print(f"\n최종 선택 모델: {best_model_name}")

print("\n[Top SNP]")

display(
    snp_result.head(10)
)

print("\n[유사 개체 분석]")

display(
    similarity_summary.round(2)
)

print("\nPCA 설명 분산")

print(
    f"PC1: "
    f"{pca.explained_variance_ratio_[0] * 100:.2f}%"
)

print(
    f"PC2: "
    f"{pca.explained_variance_ratio_[1] * 100:.2f}%"
)

print(
    f"PC1 + PC2: "
    f"{pca.explained_variance_ratio_.sum() * 100:.2f}%"
)


# ============================================================
# 32. 결과 파일 저장
# ============================================================

os.makedirs("results", exist_ok=True)

class_distribution.to_csv(
    "results/class_distribution.csv",
    encoding="utf-8-sig"
)

results.to_csv(
    "results/model_comparison.csv",
    index=False,
    encoding="utf-8-sig"
)

snp_result.to_csv(
    "results/snp_importance.csv",
    index=False,
    encoding="utf-8-sig"
)

pca_df.to_csv(
    "results/pca_result.csv",
    index=False,
    encoding="utf-8-sig"
)

similarity_summary.to_csv(
    "results/similarity_summary.csv",
    index=False,
    encoding="utf-8-sig"
)

class_similarity_df.to_csv(
    "results/individual_similarity.csv",
    index=False,
    encoding="utf-8-sig"
)

print("\n결과 파일 저장 완료")
print("results 폴더를 확인하세요.")
```
